```python
import os
import tempfile
import numpy as np
import torch
import av
import cv2
from datasets import load_dataset


DATASET_REPO = "nexar-ai/nexar_collision_prediction"

OUTPUT_CACHE_DIR = "./cached_features"

os.makedirs(OUTPUT_CACHE_DIR, exist_ok=True)

#Streaming LOOP

print("starting process...")

ds = load_dataset(DATASET_REPO, split="train", streaming=True)

for i, example in enumerate(ds):

    video_field = example["video"]

    label = example.get("label", 0)

    clip_id = example.get("id", f"video_{i:05d}")

    # Create temp file
    
    fd, tmp_video_path = tempfile.mkstemp(suffix=".mp4")

    os.close(fd)

    try:

        #Download
        
        if isinstance(video_field, dict):

            raw = video_field.get("bytes")

            path = video_field.get("path")

            if raw:

                with open(tmp_video_path, "wb") as f:

                    f.write(raw)

            elif path is not None:

                with open(path, "rb") as src, open(tmp_video_path, "wb") as dst:

                    dst.write(src.read())

        elif hasattr(video_field, "read"):

            with open(tmp_video_path, "wb") as f:

                f.write(video_field.read())

        elif isinstance(video_field, (bytes, bytearray)):

            with open(tmp_video_path, "wb") as f:

                f.write(video_field)

        elif isinstance(video_field, str):

            with open(video_field, "rb") as src, open(tmp_video_path, "wb") as dst:

                dst.write(src.read())

        #find tmp_video_path

        #Model

        #save_path_npy =
        #np.save(save_path_npy, sampled_frames)

    finally:

        #Delete tmp video

        if os.path.exists(tmp_video_path):
            os.remove(tmp_video_path)

print("Done !")
```
